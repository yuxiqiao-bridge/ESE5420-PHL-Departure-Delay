# ESE 5420 — Reproducible PHL baseline checkpoint
Uses the original 2025 BTS PHL-origin flight records (including cancelled flights), provided in `PHL_2025_raw_departures.csv`. The script filters cancelled flights, trains on January–August, tunes on September–October, and **never scores November–December**.

```python
# Download the script and dataset automatically when running in Google Colab.
from pathlib import Path
import os
import subprocess

repo_url = 'https://github.com/yuxiqiao-bridge/ESE5420-PHL-Departure-Delay.git'
repo_dir = Path('/content/ESE5420-PHL-Departure-Delay') if Path('/content').exists() else Path.cwd() / 'ESE5420-PHL-Departure-Delay'

if not (Path('PHL_checkpoint_auto.py').is_file() and Path('PHL_2025_raw_departures.csv').is_file()):
    if not repo_dir.exists():
        subprocess.run(['git', 'clone', repo_url, str(repo_dir)], check=True)
    os.chdir(repo_dir)

for name in ('PHL_checkpoint_auto.py', 'PHL_2025_raw_departures.csv'):
    assert Path(name).is_file(), f'Missing {name} in {Path.cwd()}'
print('Ready to run from:', Path.cwd())
```
